```python
query = """
SELECT Category,
       SUM(Sales) AS Total_Sales
FROM superstore
GROUP BY Category
HAVING SUM(Sales) > 100000;
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT
    "Customer Name",
    SUM(Sales) AS Total_Spending,

    CASE
        WHEN SUM(Sales) >= 5000 THEN 'High'
        WHEN SUM(Sales) >= 2000 THEN 'Medium'
        ELSE 'Low'
    END AS Customer_Segment

FROM superstore
GROUP BY "Customer Name"
ORDER BY Total_Spending DESC;
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT
    strftime('%Y-%m', "Order Date") AS Month,
    SUM(Sales) AS Total_Sales
FROM superstore
GROUP BY Month
ORDER BY Month;
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT
    AVG(Sales) AS Average_Sales,
    MIN(Sales) AS Minimum_Sales,
    MAX(Sales) AS Maximum_Sales
FROM superstore;
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT
    COUNT(*) AS Total_Orders,
    SUM(Sales) AS Total_Sales,
    SUM(Quantity) AS Total_Quantity,
    SUM(Profit) AS Total_Profit,
    AVG(Sales) AS Average_Sales
FROM superstore;
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT
    "Customer Name",
    SUM(Sales) AS Total_Sales
FROM superstore
GROUP BY "Customer Name"
HAVING SUM(Sales) > (
    SELECT AVG(Customer_Total)
    FROM (
        SELECT SUM(Sales) AS Customer_Total
        FROM superstore
        GROUP BY "Customer Name"
    )
)
ORDER BY Total_Sales DESC;
"""

pd.read_sql(query, conn)
```

```python
query = """
WITH region_sales AS (
    SELECT
        Region,
        SUM(Sales) AS Total_Sales
    FROM superstore
    GROUP BY Region
)

SELECT *
FROM region_sales
ORDER BY Total_Sales DESC;
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT
    "Customer Name",
    SUM(Sales) AS Total_Sales,

    RANK() OVER (
        ORDER BY SUM(Sales) DESC
    ) AS Sales_Rank

FROM superstore
GROUP BY "Customer Name"
ORDER BY Sales_Rank;
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT
    "Customer Name",
    SUM(Sales) AS Total_Sales,

    ROW_NUMBER() OVER (
        ORDER BY SUM(Sales) DESC
    ) AS Row_Number

FROM superstore
GROUP BY "Customer Name"
ORDER BY Row_Number;
"""

pd.read_sql(query, conn)
```

```python
query = """
WITH monthly_sales AS (
    SELECT
        strftime('%Y-%m', "Order Date") AS Month,
        SUM(Sales) AS Total_Sales
    FROM superstore
    GROUP BY Month
)

SELECT
    Month,
    Total_Sales,

    LAG(Total_Sales) OVER (
        ORDER BY Month
    ) AS Previous_Month_Sales

FROM monthly_sales
ORDER BY Month;
"""

pd.read_sql(query, conn)
```

```python
query = """
WITH monthly_sales AS (
    SELECT
        strftime('%Y-%m', "Order Date") AS Month,
        SUM(Sales) AS Total_Sales
    FROM superstore
    GROUP BY Month
)

SELECT
    Month,
    Total_Sales,

    LEAD(Total_Sales) OVER (
        ORDER BY Month
    ) AS Next_Month_Sales

FROM monthly_sales
ORDER BY Month;
"""

pd.read_sql(query, conn)
```

```python
query = """
CREATE VIEW IF NOT EXISTS category_summary AS

SELECT
    Category,
    SUM(Sales) AS Total_Sales,
    SUM(Profit) AS Total_Profit,
    SUM(Quantity) AS Total_Quantity

FROM superstore

GROUP BY Category;
"""

conn.execute(query)
conn.commit()
```

```python
query = """
SELECT *
FROM category_summary
ORDER BY Total_Sales DESC;
"""

pd.read_sql(query, conn)
```
